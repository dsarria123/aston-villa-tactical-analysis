## Aston Villa's Tactical Identity Under Unai Emery

### 1. Overview

This project investigates Aston Villa's tactical identity under Unai Emery using historical English Premier League football data.

The data collection stage obtains match-level and event-level information to provide a foundation for the project's subsequent analysis.

### Research Question

Can Aston Villa's tactical identity under Unai Emery be identified from event-level data, and how does its predicted effectiveness compare against different Premier League playing styles?

### Data Collection Objectives

1. Retrieve historical Premier League match-level and event-level data.
2. Collect information across multiple Premier League seasons.
3. Assess data availability and consistency.
4. Examine the structure and variables of the collected datasets.
5. Save the original datasets locally in a reproducible format.

### Dataset Scope

The collection covers seven Premier League seasons, from 2019/20 to 2025/26. Particular attention will be given to 2022/23, 2023/24 and 2024/25, which cover Aston Villa's seasons under Unai Emery.

### 2. Data Sources

This project uses the [OpenPlay Python package](https://github.com/nlbair/openplay) to retrieve historical English Premier League match-level and event-level data.

The repository provides documentation and instructions for accessing competition, season, match and event information.

Two datasets are collected:

- **Match-level data:** Includes match dates, teams, scores, managers, formations and other fixture information.
- **Event-level data:** Includes individual match actions such as passes, shots, tackles, interceptions, clearances and ball recoveries.

Both datasets are relevant to the research because they provide information about Premier League fixtures and the individual actions recorded during those matches.

**Data attribution:** The OpenPlay repository identifies its data under a Creative Commons Attribution 4.0 International (CC BY 4.0) licence. OpenPlay and its creator will be acknowledged in the project.

### 3. Data Collection Methods

Data collection is performed in Python using the OpenPlay package. Pandas is used to inspect and organise the retrieved datasets, while pathlib manages file paths.

The collection process involves:

1. Retrieving match-level data for each selected season.
2. Retrieving the corresponding event-level data.
3. Saving match records as CSV files.
4. Saving event records as compressed CSV files.
5. Checking the collected datasets for consistency and completeness.

Data is collected and saved separately by season. This allows the original files to be managed and inspected individually.

**Tools used:** Python, OpenPlay, Pandas and pathlib.

```python
#Import libraries
import openplay as op
import pandas as pd
from pathlib import Path
```

### 4. Collecting Match-Level Data

The match-level dataset contains one record for each retrieved Premier League fixture. Variables include match identifiers, dates, competing teams, scores, managers and formations.

The data is retrieved separately for each season and saved as individual CSV files in `data/raw-data`.

```python
#Select the seven Premier League seasons
seasons = ["2019/20", "2020/21", "2021/22", "2022/23", "2023/24", "2024/25", "2025/26"]

#Set the directory for raw data
raw_data_dir = Path("../data/raw-data")
raw_data_dir.mkdir(parents=True, exist_ok=True)

#Collect and save match-level data for each season
match_data = {}

for season in seasons:
    season_file = season.replace("/", "_")
    file_path = raw_data_dir / f"premier_league_matches_{season_file}.csv"

    if file_path.exists():
        print(f"{season}: File already exists — skipping download.")
        match_data[season] = pd.read_csv(file_path)
    else:
        print(f"Collecting matches for {season}...")
        df = op.matches("eng", season)
        df.to_csv(file_path, index=False)
        match_data[season] = df
        print(f"Saved {len(df)} matches.")

print("Match-level collection complete.")
```

### 5. Collecting Event-Level Data

The event-level dataset records individual actions during Premier League matches. Each row represents an event, such as a pass, shot, tackle, interception or ball recovery.

Because these datasets contain substantially more records than the match-level data, they are saved as compressed CSV files (`.csv.gz`) to reduce storage requirements.

```python
#Collect and save event-level data for each season
event_data = {}

for season in seasons:
    season_file = season.replace("/", "_")

    file_path = f"../data/raw-data/premier_league_events_{season_file}.csv.gz"

    if Path(file_path).exists():
        print(f"{season}: File already exists — skipping download.")
        event_data[season] = pd.read_csv(file_path)
    else:
        print(f"Collecting events for {season}...")
        df = op.events("eng", season)
        df.to_csv(file_path, index=False, compression="gzip")
        event_data[season] = df
        print(f"Saved {len(df):,} events.")

print("Event-level collection complete.")
```

### 6. Collection Results

The collection process produced match-level and event-level datasets for seven Premier League seasons, covering 2019/20 to 2025/26.

The following summary displays the number of match records, event records and columns collected for each season.

```python
#Summarise the collected data for each season
collection_summary = []

for season in seasons:
    matches_df = match_data[season]
    events_df = event_data[season]

    collection_summary.append({
        "Season": season,
        "Matches": len(matches_df),
        "Match Columns": len(matches_df.columns),
        "Events": len(events_df),
        "Event Columns": len(events_df.columns),
        "Unique Matches in Events": events_df["match_id"].nunique()
    })

summary_df = pd.DataFrame(collection_summary)

display(summary_df)
```

### 7. Data Validation

The collected datasets are checked to assess consistency between match-level and event-level records.

The validation checks identify duplicate match identifiers, event records without a corresponding match record, and matches without associated event records.

The `match_id` field is used to compare the two datasets.

```python
#Check that match IDs are consistent across the match and event datasets
validation_results = []

for season in seasons:
    matches_df = match_data[season]
    events_df = event_data[season]

    match_ids = set(matches_df["match_id"].dropna())
    event_match_ids = set(events_df["match_id"].dropna())

    validation_results.append({
        "Season": season,
        "Duplicate Match IDs": matches_df["match_id"].duplicated().sum(),
        "Events Without Match Record": len(event_match_ids - match_ids),
        "Matches Without Events": len(match_ids - event_match_ids)
    })

validation_df = pd.DataFrame(validation_results)

display(validation_df)
```

The validation checks found no duplicate match identifiers, no event records without a corresponding match record, and no match records without associated event data.

However, the collection summary shows that 2019/20 contains 379 matches and 2021/22 contains 377 matches, rather than the expected 380. OpenPlay also reported coverage warnings for these seasons.

The event datasets contain between 149 and 152 columns across seasons, indicating some variation in the available variables.
